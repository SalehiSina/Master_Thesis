## UNI

### Crop 1

```python
edata = sc.read_h5ad("/content/drive/MyDrive/Thesis/Projects/Data/Breast_Cancer/FMs_1/UNI.h5ad")
M = edata.obsm['UNI']
ad = adata.copy()
ad.obsm["morpho"] = M
```

```python
edata.obs_names = [cid[61:] for cid in edata.obs['cell_id']]
common_cells = ad.obs_names.intersection(edata.obs_names)
ad.obsm['morpho'][ad.obs_names.get_indexer(common_cells)] = \
    edata.obsm['UNI'][edata.obs_names.get_indexer(common_cells)]
```

```python
mask = ad.obs['annotation'].isin(['DCIS_1', 'DCIS_2', 'Invasive'])
ad = ad[mask]
# Build neighbors using attention space
sc.pp.neighbors(ad, use_rep="morpho")

# Compute UMAPac
sc.tl.umap(ad)

# Plot
sc.pl.umap(ad, color="annotation", frameon=False)
```

```python
#ad.obsm["morpho"] = normalize(ad.obsm["morpho"])
ad = ad
# Build neighbors using attention space
sc.pp.neighbors(ad, use_rep="morpho")

# Compute UMAPac
sc.tl.umap(ad)

# Plot
sc.pl.umap(ad, color="cell_type", frameon=False)
```
