## Training Plots

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df_GNN_BC = pd.read_csv("./Data/Train_Results/results_G_BC.csv")
df_GNN_MB = pd.read_csv("./Data/Train_Results/results_G_MB.csv")
df_SB_BC = pd.read_csv("./Data/Train_Results/results_BC.csv")
df_SB_MB = pd.read_csv("./Data/Train_Results/results_MB.csv")
```

```python
df_GNN_MB["Model_Name"] = df_GNN_MB["Model_Name"].replace({
    "Multi-Modal GNN (UNI)": "GNN (UNI)",
    "Single-Modal GNN (ST)": "GNN (No Morpho)",
    "Multi-Modal GNN (NOISE)": "GNN (NOISE)"
})
```

```python
df_SB_MB["Model_Name"] = df_SB_MB["Model_Name"].replace({
    "Multi-Modal Steamboat (UNI)": "Modified Steamboat (UNI)",
    "Single-Modal Steamboat (ST)": "Modified Steamboat (No Morpho)",
    "Multi-Modal Steamboat (NOISE)": "Modified Steamboat (NOISE)"
})
df_SB_BC["Model_Name"] = df_SB_BC["Model_Name"].replace({
    "Multi-Modal Steamboat (UNI)": "Modified Steamboat (UNI)",
    "Single-Modal Steamboat (ST)": "Modified Steamboat (No Morpho)",
    "Multi-Modal Steamboat (NOISE)": "Modified Steamboat (NOISE)"
})
```

### GNN

```python
palette = {
    "GNN (UNI)": "tab:blue",
    "GNN (No Morpho)": "tab:orange",
    "GNN (NOISE)": "tab:green"
}

sns.lineplot(data=df_GNN_BC, x='MaskingRate', y='loss', hue='Model_Name',marker='o', palette=palette)
sns.despine()
plt.legend(loc='upper left', frameon=False)
plt.ylim([0.05,0.19])
plt.savefig("GNN_BC_loss.pdf", format="pdf", bbox_inches="tight")
plt.show()
```

```python
result_GNN_BC = (
    df_GNN_BC.loc[
        df_GNN_BC.groupby(['Model_Name', 'MaskingRate'])['loss'].idxmin(),
        ['Model_Name', 'MaskingRate', 'alpha']
    ]
    .reset_index(drop=True)
)
```
